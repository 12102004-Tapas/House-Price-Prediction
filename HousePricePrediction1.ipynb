```python
# Importing the dependancies 
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
house_price_dataframe = pd.read_csv('housing.csv')
```

```python
print(house_price_dataframe)
```

```python
columns = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 
           'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'PRICE']

house_price_dataframe = pd.read_csv('housing.csv', header=None, names=columns, sep = '\s+')
```

```python
# Print the first 5 rows of dataframe
house_price_dataframe.head()
```

```python
# Checking the rows and coloumnns
house_price_dataframe.shape
```

```python
# Checking the missing values 
house_price_dataframe.isnull().sum()
```

```python
# Statistical measure of the dataset
house_price_dataframe.describe()
```

```python
# Understanding the correlation between varius features in the dataset-> 1-Positive corealtion 2-Negative correlation
correlation = house_price_dataframe.corr()
```

```python
# Constructing a heatmap for understand the correlation
plt.figure(figsize = (10,10))
sns.heatmap(correlation, cbar = True, square = True, fmt = '.1f', annot = True, annot_kws={'size':8}, cmap = 'Blues')
```

```python
# Spliting the data and target
X = house_price_dataframe.drop(['PRICE'], axis=1)
Y = house_price_dataframe['PRICE']
```

```python
print(X)
print(Y)
```

```python
# Spliting the data into training the data and test the data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
# Model training
# XGBoost Regresser
```

```python
# Loading the model
# NEW - tuned model
model = XGBRegressor(
    n_estimators=300,        # reduced from 500                           (number of trees to build)
    learning_rate=0.03,      # slower learning                            (how fast the model learns (slower = more accurate))
    max_depth=3,             # reduced from 4 (simpler trees)             (how deep each tree can grow)
    min_child_weight=8,      # increased from 5 (stricter)                (controls overfitting)
    subsample=0.7,           # reduced from 0.8                           (uses 80% of data per tree (prevents overfitting))
    colsample_bytree=0.7,    # reduced from 0.8                           (uses 80% of features per tree)
    reg_alpha=0.5,           # increased from 0.1 (more regularization)   (L1 regularization (reduces overfitting))
    reg_lambda=2.0,          # increased from 1.5 (more regularization)   (L2 regularization (reduces overfitting))
    gamma=0.2,               # increased from 0.1 (harder to split)       (minimum loss reduction to make a split)
    random_state=42          # reproducible results every run
)
```

```python
# Training the model with X_train
model.fit(X_train, Y_train)
```

```python
# Evaluation
# Prediction on training data
```

```python
# Accuracy for prediction on training data
training_data_prediction = model.predict(X_train)
```

```python
print(training_data_prediction)
```

```python
# R squared error
score_1 = metrics.r2_score(Y_train, training_data_prediction)

# Mean absolute error
score_2 = metrics.mean_absolute_error(Y_train, training_data_prediction)

print('R squared error :', score_1)
print('Mean absolute error :', score_2)
```

```python
# Visualizing the actual prices and predicted prices
plt.scatter(Y_train, training_data_prediction)
plt.xlabel('Actual Prices')
plt.ylabel('Mean Absolute Error')
plt.title('Actual Prices VS Predicted Prices')
plt.show()
```

```python
# Prediction on test data
```

```python
# Accuracy for prediction on test data
test_data_prediction = model.predict(X_test)
```

```python
# R squared error
score_1 = metrics.r2_score(Y_test, test_data_prediction)

# Mean absolute error
score_2 = metrics.mean_absolute_error(Y_test, test_data_prediction)

print('R squared error :', score_1)
print('Mean absolute error :', score_2)
```

```python
# Training evaluation
training_data_prediction = model.predict(X_train)
score_1 = metrics.r2_score(Y_train, training_data_prediction)
score_2 = metrics.mean_absolute_error(Y_train, training_data_prediction)
print('Training R squared error :', score_1)
print('Training Mean absolute error :', score_2)

# Test evaluation
test_data_prediction = model.predict(X_test)
score_1 = metrics.r2_score(Y_test, test_data_prediction)
score_2 = metrics.mean_absolute_error(Y_test, test_data_prediction)
print('Test R squared error :', score_1)
print('Test Mean absolute error :', score_2)
```
